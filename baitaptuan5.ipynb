```python
#W5A1Viết hàm truyền vào hai số nguyên a, b. Trả về số lớn hơn trong hai số. def Max_Of_Two( a, b) -> 
def solonhon(a,b):
    if a>b:
        print(a)
    else:        print(b)
a,b=map(int,input("Nhập hai số nguyên: ").split())
solonhon(a,b)
```

```python
#W5A2 Viết hàm hoán đổi hai số nguyên. def swap(a, b) -> Tuple[int, int]: 
def swap(a,b):
    a,b=b,a
    return a,b
a,b=map(int,input("Nhập hai số nguyên: ").split())
swap(a,b)
```

```python
#W5A3 Viết hàm kiểm tra một số có phải là số nguyên tố hay không?
import math
def kiemtrasnt(n):
    if n<2:
        return False
    for i in range(2,int(math.sqrt(n))+1):
        if n%i==0:
            return print("Không phải là số nguyên tố")
        return print("Là số nguyên tố")
n=int(input("Nhập một số nguyên: "))
kiemtrasnt(n)
```

```python
#W5A4 VIết hàm kiểm tra một số có phải số hoàn hảo hay không, xử lý trên dữ liệu kiểu số nguyên. 
def sohoanhao(n):
    tong=0
    for i in range(1,n):
        if n%i==0:
            tong+=i
    if tong==n:
        print(f"{n} là số hoàn hảo")
    else:
        print(f"{n} không phải là số hoàn hảo")
n=int(input("Nhập một số nguyên: "))
sohoanhao(n)
```

```python
#W5A5Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và trả về vị trí của phần tử đầu tiên trong list 
# có giá trị lớn hơn k. Nếu không tìm thấy phần tử nào thỏa mãn điều kiện, trả về -1.
def timvitri(list,k):
    for i in range(len(list)):
        if list[i]>k:
            return i
    return -1
list=[int(x) for x in input("Nhập một list số nguyên (cách nhau bằng dấu cách): ").split()]
k=int(input("Nhập một số nguyên dương k: "))
timvitri(list,k)
```

```python
#W5A6 Viết hàm tính giai thừa. Nhập vào số nguyên n, trả về giai thừa của n (n!) 
tich=1
def giaithua(n):
    if n==0 or n==1:
        return 1
    for i in range(2,n+1):
        tich*=i
    print(f"Giai thừa của {n} là: {tich}")
n=int(input("Nhập một số nguyên: "))
giaithua(n)
    
```
