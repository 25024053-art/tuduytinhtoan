## Bài W7A18: Sắp xếp có điều kiện: Số chẵn đứng

```python
# lam theo huong dan tren lop
def sort_cond(a):
    evens = sorted([x for x in a if x % 2 == 0])
    odds = sorted([x for x in a if x % 2 != 0], reverse=True)
    return evens + odds

a = [32, 4, 15, 123, 24, 12]
print('Ket qua test:', 'Chẵn tăng/Lẻ giảm:', sort_cond(a))
```

## Bài W7A22: Cho N tên sinh viên gồm cả họ và tên,

```python
# lam theo huong dan tren lop

def best_name(names):
    n = [x.split()[-1] for x in names]
    c = Counter(n)
    return min([k for k, v in c.items() if v == max(c.values())])

names = ['Bùi Bài Bình', 'Nguyễn Tất Bình', 'Nguyễn Mạnh Cường']
print('Ket qua test:', 'Tên hay gặp nhất:', best_name(names))
```

## Bài W7A24: Cho xâu kí tự, xóa các từ giống nhau liền

```python
# lam theo huong dan tren lop
def rem_dup(s):
    w = s.split()
    return ' '.join([w[i] for i in range(len(w)) if i==0 or w[i]!=w[i-1]])

print('Ket qua test:', 'Kết quả:', rem_dup('Toi di di hoc, rat vui khi di hoc hoc'))
```

## Bài W7A25: Tìm số có đúng k chữ số lớn nhất có thể

```python
# lam theo huong dan tren lop
def max_k(s, k):
    return max(int(s[i:i+k]) for i in range(len(s)-k+1))

print('Ket qua test:', 'Max k:', max_k('454878434354', 5))
```

## Bài W7A26: Tách các số có trong một đoạn văn bản,

## Bài W7A27: Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra

```python
# lam theo huong dan tren lop
def valid(s):
    st = []
    map = {')': '(', '}': '{', ']': '['}
    for c in s:
        if c in map.values(): st.append(c)
        elif c in map:
            if not st or st.pop() != map[c]: return False
    return not st

print('Ket qua test:', 'Valid:', valid('({}{})'))
```

## Bài W7A29: Tìm Dãy con tăng dần liên tục dài nhất

```python
# lam theo huong dan tren lop
def LCI(a):
    m, c = [a[0]], [a[0]]
    for i in range(1, len(a)):
        if a[i] > a[i-1]: c.append(a[i])
        else: c = [a[i]]
        if len(c) > len(m): m = c
    return m
print('Ket qua test:', 'Dãy con liên tục:', LCI([1,2,3,4,1,2,3,4,9,10,12]))
```

## Bài W7A30: Tìm Dãy con tăng dần KHÔNG liên tục

```python
# lam theo huong dan tren lop
def LIS(a):
    dp = [1]*len(a)
    for i in range(len(a)):
        for j in range(i):
            if a[i] > a[j]: dp[i] = max(dp[i], dp[j]+1)
    return max(dp)
print('Ket qua test:', 'LIS:', LIS([1,2,3,4,5,6,4,6,7,9,10,12]))
```
