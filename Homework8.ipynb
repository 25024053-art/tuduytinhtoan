## Bài W8A16: Yêu cầu

```python
class Shape:
    def area(self): pass
class Rectangle(Shape):
    def __init__(self, w, h): self.w, self.h = w, h
    def area(self): return self.w * self.h
class Circle(Shape):
    def __init__(self, r): self.r = r
    def area(self): return 3.14 * self.r**2

# Example:
r = Rectangle(5, 10); print(f'Rectangle area: {r.area():.2f}')
```

## Bài W8A17: Yêu cầu

```python
class Vehicle:
    def __init__(self, b, s): self.b, self.s = b, s
    def info(self): return f'{self.b}-{self.s}'
class Car(Vehicle):
    def __init__(self, b, s, seats): super().__init__(b, s); self.seats = seats
    def info(self): return f'{self.b}-{self.s}-{self.seats}'

# Example:
c = Car('Toyota', 120, 5); print(c.info())
```

## Bài W8A18: Yêu cầu

```python
class Polygon:
    def perimeter(self): pass
class Triangle(Polygon):
    def __init__(self, a, b, c): self.a, self.b, self.c = a, b, c
    def perimeter(self): return self.a + self.b + self.c

# Example:
t = Triangle(3, 4, 5); print(t.perimeter())
```

## Bài W8A19: Yêu cầu

```python
class Person:
    def __init__(self, n, a): self.n, self.a = n, a
    def describe(self): return f'{self.n} {self.a}'
class Student(Person):
    def __init__(self, n, a, sch): super().__init__(n, a); self.sch = sch
    def describe(self): return f'{self.n} {self.a} {self.sch}'

# Example:
s = Student('Nam', 20, 'UET'); print(s.describe())
```

## Bài W8A20: Yêu cầu

```python
class Account:
    def __init__(self, b): self.b = b
    def deposit(self, x): self.b += x
    def get_balance(self): return self.b
class SavingAccount(Account):
    def __init__(self, b, r): super().__init__(b); self.r = r
    def add_interest(self): self.b += self.b * self.r

# Example:
s = SavingAccount(1000, 0.05); s.add_interest(); print(f'{s.get_balance():.2f}')
```

## Bài W8A22: Yêu cầu

```python
class Matrix2x2:
    def __init__(self, a, b, c, d): self.a, self.b, self.c, self.d = a, b, c, d
    def trace(self): return self.a + self.d
    def determinant(self): return self.a * self.d - self.b * self.c

# Example:
m = Matrix2x2(1, 2, 3, 4); print(m.trace(), m.determinant())
```

## Bài W8A27: Yêu cầu

```python
class Product:
    def __init__(self, n, p): self.n, self.p = n, p
class DiscountProduct(Product):
    def __init__(self, n, p, d): super().__init__(n, p); self.d = d
    def final_price(self): return self.p * (100 - self.d) / 100

# Example:
d = DiscountProduct('Pen', 100, 20); print(f'{d.final_price():.2f}')
```

## Bài W8A28: Yêu cầu

```python
class Counter:
    def __init__(self): self.count = 0
    def inc(self): self.count += 1
    def dec(self): self.count -= 1
    def value(self): return self.count

# Example:
c = Counter(); c.inc(); print(c.value())
```

## Bài W8A29: Yêu cầu

```python
class Time:
    def __init__(self, h, m, s): self.h, self.m, self.s = h, m, s
    def to_seconds(self): return self.h * 3600 + self.m * 60 + self.s

# Example:
t = Time(1, 1, 1); print(t.to_seconds())
```

## Bài W8A30: Yêu cầu

```python
class Stack:
    def __init__(self): self.items = []
    def push(self, x): self.items.append(x)
    def pop(self): return self.items.pop()
    def top(self): return self.items[-1]

# Example:
s = Stack(); s.push(1); print(s.top())
```
