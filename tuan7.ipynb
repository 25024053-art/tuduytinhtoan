
### W7A1: Tìm kiếm nhị phân (Binary Search)
def binary_search(arr, left, right, target):
    if left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] > target:
            return binary_search(arr, left, mid - 1, target)
        else:
            return binary_search(arr, mid + 1, right, target)
    return -1
```

### W7A2: Đếm số lần xuất hiện (Count Occurrences)
def count_occurrences(arr, x) -> int:
    return arr.count(x)

### W7A3: Bubble Sort và Selection Sort
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr

def selection_sort(arr):
    for i in range(len(arr)):
        min_idx = i
        for j in range(i+1, len(arr)):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr
```

### W7A5: Đếm cặp tổng bằng X

def count_pairs(arr, X):
    count = 0
    for i in range(len(arr)):
        for j in range(i + 1, len(arr)):
            if arr[i] + arr[j] == X:
                count += 1
    return count
```

### W7A6: Làm phẳng danh sách lồng nhau (Flatten List)

def flatten_list(nested_list):
    flat = []
    for sublist in nested_list:
        for item in sublist:
            flat.append(item)
    return flat
```

def length_of_lis(nums):
    if not nums: return 0
    dp = [1] * len(nums)
    for i in range(1, len(nums)):
        for j in range(i):
            if nums[i] > nums[j]:
                dp[i] = max(dp[i], dp[j] + 1)
    return max(dp)
```

### W7A8: Tìm khoảng thời gian ngắn nhất bao quanh truy vấn
def solve_intervals(intervals, queries):
    results = []
    for q in queries:
        min_len = float('inf')
        found = False
        for l, r in intervals:
            if l <= q <= r:
                min_len = min(min_len, r - l)
                found = True
        results.append(min_len if found else -1)
    return results
```

### W7A12: Sắp xếp từ điển
```python
def dictionary_order(words):
    indexed_words = sorted(enumerate(words), key=lambda x: x[1])
    labels = ['A', 'B', 'C', 'D']
    return " ".join([labels[i] for i, word in indexed_words])
```

### W7A13: Dãy hài hòa con dài nhất
```python
from collections collections Counter
def longest_harmonious_subsequence(arr):
    count = Counter(arr)
    max_len = 0
    for x in count:
        if x + 1 in count:
            max_len = max(max_len, count[x] + count[x+1])
    return max_len
```

### W7A14: Tìm tất cả số 7 (Vị trí giảm dần)
```python
def find_sevens(n, arr):
    indices = [i for i, x in enumerate(arr) if x == 7]
    if not indices: return "Not found"
    return " ".join(map(str, sorted(indices, reverse=True)))
```

### W7A15: Tìm người ngồi cạnh Nemo
```python
def find_neighbors(n, guests):
    try:
        idx = guests.index("Nemo")
        prev_guest = guests[idx-1] if idx > 0 else "None"
        next_guest = guests[idx+1] if idx < n-1 else "None"
        return f"{prev_guest} and {next_guest}"
    except ValueError:
        return "Nemo not found"
```
