2 Yêu cầu: Cho số nguyên x, trả về True nếu tồn tại số nguyên n ≥ 1 sao cho x = 1 + 2 + ... + n, ngược lại trả về False.
Khuyến khích dùng vòng lặp đơn giản để cộng dồn tổng. 


```python
x=int(input())
def so(x):
    if x < 1:
        return False
    tong = 0
    n = 1
    while tong < x:
        tong += n
        if tong == x:
            return True
        n += 1
    return False
print(so(x))
```

- d < 0 → []
- d == 0 → [ -b / (2a) ]
- d > 0 → hai nghiệm [r1, r2] với r1 <= r2.


```python
import sympy as sp
x=sp.symbols('x')
a,b,c=map(int,input().split())
pt=a*x*x+b*x+c
if a==0:
    if b==0:
        if c==0:
            print("không có nghiệm hữu hạn duy nhất")
        if c!=0:
            print("vô nghiệm")
    if b!=0:
        nghiem=-c/b
        print(nghiem)
if a!=0:
    nghiem=sp.solve(pt,x)
    print(nghiem)  
```
