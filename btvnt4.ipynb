```python
n=int(input())
def uoc(n):
    if n<0:
        return False
    for i in range(1,n+1):
        if n%i==0:
         print(i,end=' ')
def kiem_tra_snt(n):
    if n==1:
        return False
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            return False
    return True
def tim_uoc_nguyen_to_lon_nhat(n):
    uln = 0
    for i in range(2,n+1):
        if n%i==0:
            if kiem_tra_snt(i):
                uln = i
    return uln
print("\n Ước nguyên tố lớn nhất là:",tim_uoc_nguyen_to_lon_nhat(n))
```
