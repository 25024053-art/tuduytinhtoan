## Bài 1: Tìm số lớn nhất của 1 list
Viết chương trình nhập vào một danh sách các số nguyên, tìm và in ra số lớn nhất trong danh sách.

```python
def tim_max(lst):
    if not lst:
        return "Danh sách rỗng"
    
    max_val = lst[0]
    for num in lst:
        if num > max_val:
            max_val = num
    return max_val

# Thử nghiệm
lst1 = [1, 5, 3, 9, 2, 7]
print("Danh sách:", lst1)
print("Số lớn nhất:", tim_max(lst1))
```

## Bài 2: Tìm số nhỏ nhất của 1 list
Viết chương trình nhập vào một danh sách các số nguyên, tìm và in ra số nhỏ nhất trong danh sách.

```python
def tim_min(lst):
    if not lst:
        return "Danh sách rỗng"
    
    min_val = lst[0]
    for num in lst:
        if num < min_val:
            min_val = num
    return min_val

# Thử nghiệm
lst2 = [4, 5, 1, 9, 2, 7]
print("Danh sách:", lst2)
print("Số nhỏ nhất:", tim_min(lst2))
```

## Bài 2b: Tìm số nhỏ nhất của 1 list và vị trí của nó
Viết chương trình nhập vào một danh sách các số nguyên, tìm số nhỏ nhất và vị trí đầu tiên của nó (nếu có nhiều số nhỏ nhất bằng nhau). Nếu danh sách rỗng, in ra "-1".

```python
def tim_min_va_vi_tri(lst):
    if not lst:
        return -1
    
    min_val = lst[0]
    min_idx = 0
    
    for i in range(1, len(lst)):
        if lst[i] < min_val:
            min_val = lst[i]
            min_idx = i
            
    return min_val, min_idx

# Thử nghiệm
lst2b = [5, 3, 2, 8, 2, 9]
result = tim_min_va_vi_tri(lst2b)
if result == -1:
    print("-1")
else:
    val, idx = result
    print(f"Danh sách: {lst2b}")
    print(f"Số nhỏ nhất là {val} tại vị trí đầu tiên là {idx}")
```

## Bài 3: Tìm giá trị trung bình của 1 list
Viết chương trình nhập vào một danh sách các số nguyên, tính và in ra giá trị trung bình (so thuc) cua danh sach. Nếu danh sách rỗng, in ra "0".

```python
def tinh_trung_binh(lst):
    if not lst:
        return 0
    
    tong = sum(lst)
    return tong / len(lst)

# Thử nghiệm
lst3 = [1, 2, 3, 4, 5, 6]
print("Danh sách:", lst3)
print("Giá trị trung bình:", tinh_trung_binh(lst3))
```

## Bài 4: Kiểm tra một list số nguyên có phải là đối xứng hay không
Viết chương trình nhập vào một danh sách các số nguyên, kiểm tra xem danh sách có phải là đối xứng (palindrome) hay không. In ra "True" nếu là đối xứng, "False" nếu không phải.

```python
def kiem_tra_doi_xung(lst):
    n = len(lst)
    for i in range(n // 2):
        if lst[i] != lst[n - 1 - i]:
            return False
    return True

# Thử nghiệm
lst4_1 = [1, 2, 3, 2, 1]
lst4_2 = [1, 2, 3, 4, 5]
print(f"Danh sách {lst4_1} đối xứng:", kiem_tra_doi_xung(lst4_1))
print(f"Danh sách {lst4_2} đối xứng:", kiem_tra_doi_xung(lst4_2))
```

## Bài 5: Đưa tất cả các số bằng 0 trong list về đầu của list
Viết chương trình nhập vào một danh sách các số nguyên, đưa tất cả các số bằng 0 về đầu danh sách, giữ nguyên thứ tự tương ứng của các số khác.

```python
def dua_0_ve_dau(lst):
    zeros = [num for num in lst if num == 0]
    non_zeros = [num for num in lst if num != 0]
    return zeros + non_zeros

# Thử nghiệm
lst5 = [1, 0, 3, 0, 2, 0, 4]
print("Danh sách gốc:", lst5)
print("Sau khi biến đổi:", dua_0_ve_dau(lst5))
```

## Bài 6: Xóa tất cả các số giống nhau trong list
Viết chương trình nhập vào một danh sách các số nguyên, xóa tất cả các số trùng lặp, giữ nguyên thứ tự xuất hiện đầu tiên của mỗi số.

```python
def xoa_trung_lap(lst):
    ket_qua = []
    for num in lst:
        if num not in ket_qua:
            ket_qua.append(num)
    return ket_qua

# Thử nghiệm
lst6 = [1, 2, 3, 2, 1, 4, 5, 3]
print("Danh sách gốc:", lst6)
print("Sau khi xóa trùng lặp:", xoa_trung_lap(lst6))
```
