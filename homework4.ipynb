```python
n=(input())
inhoa=0
inthuong=0
so=0
for kitu in str(n):
    if kitu.isupper():
        inhoa+=1
    elif kitu.islower():
        inthuong+=1
    elif kitu.isdigit():
        so+=1
print(inhoa,inthuong,so)
```
