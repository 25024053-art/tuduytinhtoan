### Problem 4
#### Write a function that returns the count of positive integers k such that 1 <= k < n and k > (n - 1) / 2. (Example: n=9 -> 4).

```python
n=float(input("nhập số n: "))
def nhap_so(n):
    for k in range(1, int(n)):
        if k>(int(n)-1)/2 and n%2==1:
            return k-1
        if k>(int(n)-1)/2 and n%2==0:
            return k
print(nhap_so(n))
    
```
